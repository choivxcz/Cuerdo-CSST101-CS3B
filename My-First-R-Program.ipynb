# NUMERICAL OPERATORS

```r
x <- 5
y <- 6

sum <- x + y
difference <- x - y
product <- x * y
quotient <- x / y

print(sum)
print(difference)
print(product)
print(round(quotient, 2))

cat("\nSUM: ", sum, "\n")
cat("difference: ", difference, "\n")
cat("product: ", product, "\n")
cat("quotient: ", round(quotient, 2), "\n")

```

#Vectors

```r
grades <- c(76,62,85,86,90)

grades

cat("\nMEAN:",mean(grades))
cat("\nMAX",max(grades))
cat("\nMIN:",min(grades))
cat("\nSUM:",sum(grades))
cat("\nLength:",length(grades))
```
